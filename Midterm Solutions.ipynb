```python
def intensity(h, k, T, v, n, R):
    power = (h*v)/(k*T)
    frac1 = (1.76e-46) / (v**(-3)*(np.exp(power)-1))
    gff = 12 * v**(-0.1) * T**(-0.15)
    alpha = 3.7e8*(T**(-1/2))*n**2*n*v**(-3)*(1-np.exp(-power))*gff
    I = frac1*(1-np.exp(-alpha*R))
    return I

def opt_thick(h, k, T, v, n, R): # Intensity tau >> 
    power = (h*v)/(k*T)
    frac1 = (1.76e-46) / (v**(-3)*(np.exp(power)-1))
    gff = 12 * v**(-0.1) * T**(-0.15)
    alpha = 3.7e8*(T**(-1/2))*n**2*n*v**(-3)*(1-np.exp(-power))*gff
    I = frac1*(1-np.exp(-1))
    return I
```
